```python
import pandas as pd
```

```python
df = pd.read_csv("D:/SEM6/LightGCN/Dataset/ratings.csv")
print(df.head())
```

```python
df.rename(columns={
    "userId": "user",
    "movieId": "item"
}, inplace=True)
```

```python
df['user'] = df['user'].astype('category').cat.codes
df['item'] = df['item'].astype('category').cat.codes
```

```python
n_users = df['user'].nunique()
n_items = df['item'].nunique()

print(n_users, n_items)
```
